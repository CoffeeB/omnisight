# OmniSight: Multi-View Dataset Exploration & Physical Degradation Demo
This notebook demonstrates multi-view scene generation, camera viewpoint homography warping, Koschmieder atmospheric fog, and Ground Sample Distance (GSD) resampling.

```python
import os
import sys
sys.path.insert(0, os.path.abspath(".."))
import numpy as np
import matplotlib.pyplot as plt
from datasets.synthetic_generator import SyntheticSceneGenerator, CLASS_NAMES
from src.preprocessing.atmospheric import apply_koschmieder_fog, apply_rain_streaks, apply_solar_shadows
from src.preprocessing.geometry import apply_viewpoint_transformation, ViewpointConfig, ViewpointType
from src.visualization.visualizer import render_semantic_overlay

generator = SyntheticSceneGenerator(img_size=(384, 384), seed=42)
sample = generator.generate_scene()

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(sample.primary_image)
axes[0].set_title("Primary Image (Nadir / Overhead)")

overlay = render_semantic_overlay(sample.primary_image, sample.semantic_mask)
axes[1].imshow(overlay)
axes[1].set_title("Ground-Truth Semantic Mask")

if sample.auxiliary_views:
    axes[2].imshow(sample.auxiliary_views[0])
    axes[2].set_title("Auxiliary View (Side/Oblique)")
plt.tight_layout()
plt.show()
```

```python
# Atmospheric Degradations Demo
fog_img = apply_koschmieder_fog(sample.primary_image, beta=0.06)
rain_img = apply_rain_streaks(sample.primary_image, intensity=0.75)
shadow_img = apply_solar_shadows(sample.primary_image, num_shadows=4)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(fog_img)
axes[0].set_title("Koschmieder Fog (beta=0.06)")
axes[1].imshow(rain_img)
axes[1].set_title("Rain Streaks (intensity=0.75)")
axes[2].imshow(shadow_img)
axes[2].set_title("Harsh Solar Shadows")
plt.tight_layout()
plt.show()
```
